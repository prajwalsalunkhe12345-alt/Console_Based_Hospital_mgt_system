```python
#Contact Project Using Python 
#Load the modules
import os
import json
from datetime import datetime

# HOSPITAL MANAGEMENT SYSTEM

# 1. Add Patient
# 2. View Patients
# 3. Search Patient
# 4. Update Patient
# 5. Delete Patient
# 6. Patient Statistics
# 7. Exit
```

```python

PATIENT_FILE = "patients_info.json"
# -----------------------------------------------
# Loading Patient data 
# -----------------------------------------------
def load_patients():
    try:
        if not os.path.exists(PATIENT_FILE):
            return []

        with open(PATIENT_FILE, "r") as file:
            return json.load(file)

    except json.JSONDecodeError:
        print("\nError: Patient file is corrupted!")
        return []

    except OSError as e:
        print("\nFile Error:", e)
        return []

# -----------------------------------------------
# Save Patients Data  in JSON file 
# -----------------------------------------------
def save_patients(patients):
    try:
        with open(PATIENT_FILE, "w") as file:
            json.dump(patients, file, indent=4)

    except OSError as e:
        print("\nFile Error:", e)

# Automatically patient id generate 
def generate_patient_id(patients):

    if not patients:
        return "P001"

    numbers = []

    for patient in patients:
        try:
            number = int(patient["patient_id"][1:])
            numbers.append(number)

        except (ValueError, KeyError):
            continue

    if numbers:
        next_number = max(numbers) + 1
    else:
        next_number = 1

    return "P" + str(next_number).zfill(3)

# -----------------------------------------------    
# Adding Patients
# -----------------------------------------------

def add_patient():

    try:
        patients = load_patients()

        patient_id = generate_patient_id(patients)

        print("\n ---- ADD PATIENT ----")
        print(" Patient ID:", patient_id)

        # Name 
        while True:
            name = input("Enter Patient Name: ").strip()
            if name == "":
                print("Name cannot be empty!")
            elif not name.replace(" ", "").isalpha():
                print("Invalid Name! Please enter characters only.")
            else:
                break

        # Age 
        while True:
            try:
                age = int(input("Enter Age: "))
                if age <= 0:
                    print("Age must be greater than 0!")
                elif age > 80:
                    print("Please enter a valid age!")
                else:
                    break
            except ValueError:
                print("Invalid Age! Please enter numbers only.")

        # Gender
        while True:
            gender = input("Enter Gender (male/female): ").strip().lower()
            if gender in ("male", "female"):
                break
            print("Invalid Gender! Please enter male or female.")

        # Mobile 
        while True:
            mobile = input("Enter Mobile Number: ").strip()
            if not mobile.isdigit():
                print("Invalid Mobile Number! Enter Numbers.")
            elif len(mobile) != 10:
                print("Invalid Mobile Number! Enter  10 digits.")
            else:
                break

        # Disease
        while True:
            disease = input("Enter Disease: ").strip()
            if disease and disease.replace(" ", "").isalpha():
                break
            print("Invalid Disease! Please enter letters only.")
            
        # Patient Record
        patient = {
            "patient_id": patient_id,
            "name": name,
            "age": age,
            "gender": gender,
            "mobile": mobile,
            "disease": disease,
            "date": datetime.now().strftime("%d-%m-%Y")
        }

        patients.append(patient)
        save_patients(patients)
        print("\nPatient Added Successfully!")
        print("Patient ID:", patient_id)

    except Exception as e:
        print("\nError:", e)


# -----------------------------------------------
# VIEW PATIENTS
# -----------------------------------------------
def view_patients():

    try:
        patients = load_patients()

        if not patients:
            print("\nNo Patient Found!")
            return

        print("\n---- PATIENT LIST ----")

        for patient in patients:

            print("----------------------------------")
            print("Patient ID :", patient["patient_id"])
            print("Name :", patient["name"])
            print("Age :", patient["age"])
            print("Gender :", patient["gender"])
            print("Mobile :", patient["mobile"])
            print("Disease :", patient["disease"])
            print("Date:", patient["date"])
        print("----------------------------------")

    except Exception as e:
        print("\nError:", e)


# -----------------------------------------------
# SEARCH PATIENT
# -----------------------------------------------

def search_patient():

    try:
        patients = load_patients()

        if not patients:
            print("\nNo Patient Found!")
            return

        patient_id = input("Enter Patient ID to Search: " ).strip().upper()
        for patient in patients:
            if patient["patient_id"] == patient_id:
                print("\n---- PATIENT DETAILS ----")
                print("Patient ID :", patient["patient_id"])
                print("Name :", patient["name"])
                print("Age :", patient["age"])
                print("Gender :", patient["gender"])
                print("Mobile :", patient["mobile"])
                print("Disease :", patient["disease"])
                print("Date:", patient["date"])

                return
        print("\nPatient Not Found!")
    except Exception as e:
        print("\nError:", e)


# -----------------------------------------------
# UPDATE PATIENT
# -----------------------------------------------

def update_patient():
    try:
        patients = load_patients()
        if not patients:
            print("\nNo Patient Found!")
            return
        patient_id = input("Enter Patient ID to Update: ").strip().upper()
        for patient in patients:
            if patient["patient_id"] == patient_id:
                print("\n---- UPDATE PATIENT ----")

                # Name 
                while True:
                    name = input("Enter New Name: ").strip()
                    if name == "":
                        print("Name cannot be empty!")
                    elif not name.replace(" ", "").isalpha():
                        print( "Invalid Name!Please enter characters only." )
                    else:
                        break

                # Age 
                while True:
                    try:
                        age = int(input("Enter New Age: "))
                        if age <= 0:
                            print("Age must be greater than 0!")
                        elif age > 80:
                            print("Please enter a valid age!")
                        else:
                            break
                    except ValueError:
                        print("Invalid Age!Please enter numbers only.")

                # Gender

                while True:
                    gender = input("Enter New Gender (male/female): ").strip().lower()
                    if gender in ("male", "female"):
                        break
                    print("Invalid Gender! Please enter male or female.")
                # Mobile 
                while True:
                    mobile = input( "Enter New Mobile Number: ").strip()
                    if not mobile.isdigit():
                        print("Invalid Mobile Number!Enter Numbers.")
                    elif len(mobile) != 10:
                        print( "Invalid Mobile Number!Enter 10 digits." )
                    else:
                        break

                # Disease
                while True:
                    disease = input("Enter New Disease: ").strip()
                    if disease and disease.replace(" ", "").isalpha():
                        break
                    print("Invalid Disease! Please enter letters only.")
                 
                # Update Data
                patient["name"] = name
                patient["age"] = age
                patient["gender"] = gender
                patient["mobile"] = mobile
                patient["disease"] = disease

                save_patients(patients)
                print("\nPatient Updated Successfully!")
                return
        print("\nPatient Not Found!")
    except Exception as e:
        print("\nError:", e)


# -----------------------------------------------
# DELETE PATIENT
# -----------------------------------------------
def delete_patient():
    try:
        patients = load_patients()
        
        if not patients:
            print("\nNo Patient Found!")
            return  
            
        patient_id = input("Enter Patient ID to Delete: ").strip().upper()
        
        for patient in patients:
            if patient["patient_id"] == patient_id:
                print("\nPatient Name:", patient["name"])
                
                confirm = input("Are you sure you want to delete? (yes/no): ").strip().lower()
                if confirm == "yes":
                    
                    patients.remove(patient)
                    save_patients(patients)
                    print("\nPatient Deleted Successfully!")
                else:
                    print("\nDelete Operation Cancelled.")
                return
        print("\nPatient Not Found!")
    except Exception as e:
        print("\nError:", e)


# -----------------------------------------------
# PATIENT STATISTICS
# -----------------------------------------------

def patient_statistics():

    try:
        patients = load_patients()

        if not patients:
            print("\nNo Patient Records Found!")
            return

        total = len(patients)
        male = 0
        female = 0
        other = 0

        for patient in patients:

            if patient["gender"].lower() == "male":
                male += 1

            elif patient["gender"].lower() == "female":
                female += 1

            else:
                other += 1

        print("\n------------------------------------")
        print("        PATIENT STATISTICS")
        print("------------------------------------")
        print("Total Patients  :", total)
        print("Male Patients   :", male)
        print("Female Patients :", female)
        print("Other Patients  :", other)
        print("------------------------------------")

    except Exception as e:
        print("\nError:", e)


# -----------------------------------------------
# MAIN MENU
# -----------------------------------------------

def menu():

    while True:

        try:

            print("\n------------------------------------")
            print("      HOSPITAL MANAGEMENT SYSTEM")
            print("------------------------------------")
            print("1. Add Patient")
            print("2. View Patients")
            print("3. Search Patient")
            print("4. Update Patient")
            print("5. Delete Patient")
            print("6. Patient Statistics")
            print("7. Exit")
            print("------------------------------------")

            choice = input("Enter Your Choice: ").strip()

            if choice == "1":
                add_patient()

            elif choice == "2":
                view_patients()

            elif choice == "3":
                search_patient()

            elif choice == "4":
                update_patient()

            elif choice == "5":
                delete_patient()

            elif choice == "6":
                patient_statistics()

            elif choice == "7":
                print("\nThank You!")
                print("Hospital Management System...")
                break

            else:
                print(
                    "\nInvalid Choice! "
                    "Please enter a number from 1 to 7."
                )

        except KeyboardInterrupt:
            print("\n\nProgram stopped by user.")
            break

        except Exception as e:
            print("\nUnexpected Error:", e)


# -----------------------------------------------
# START PROGRAM
# -----------------------------------------------
if __name__ == "__main__":
    menu()
```
